```python
#Read ../config.yaml using yaml.safe_load and display the number of countries
```

```python
with open("../config.yaml") as f:
    config = yaml.safe_load(f)

n_countries = len(config["scope"]["countries"])
weights_sum = sum(config["risk_index"]["weights"].values())

print("Countries:  ", n_countries)
print("Weights sum:", weights_sum)

assert n_countries == 30, "Expected 30 countries"
assert abs(weights_sum - 1.0) < 1e-9, "Weights must sum to 1"
print("Config OK")
```
